```python
data = {
    'A': [1,2,3,4],
    'B': [10,20,30,40]
}
df = pd.DataFrame(data)
df
```

```python
df['Sum'] = df.apply(lambda x: x['A'] + x['B'], axis=1) # .applay() has same func. as .iterrows()
df
```

```python
df = df.map(lambda x: x * 2) # Use the .applymap on python below 3.12
df
```

```python
df['Kat'] = df.apply(lambda x: 'Tinggi' if x['Sum'] >= 30 else 'Rendah', axis=1)
df
```

```python
# Data baru
data = {
    'Nama': ['Masbro', 'MustANice', 'Poyopoyo'],
    'Umur': [20, 1000, 24],
    'Gaji': [1000000, 0, 5000]
}
df_dic = pd.DataFrame(data)
df_dic
```

```python
def kat_umur(umur):
    if umur < 25 and umur >= 0:
        return "Muda"
    elif umur >= 25:
        return "Aku sudah kecewa"

df_dic['Kategori'] = df_dic['Umur'].apply(kat_umur)
df_dic
```

```python
# REMEMBAH
# x['y']; x is rows, y is columns or value or whatever you wanna say it :v

# Alter way
def pajak(gaji, kategori):
    if kategori == "Muda":
        pajaks = gaji * 0.05
    else:
        pajaks = gaji * 0.1
    return pajaks

df_dic['Pajak_Alt'] = df_dic.apply(lambda x: pajak(x['Gaji'], x['Kategori']), axis=1)
df_dic
```

```python
# Bonus dengan fungsi
def bonus_karyawan(x):
    bonus = x['Gaji']
    return bonus * 0.1

df_dic['Bonus'] = df_dic.apply(bonus_karyawan, axis=1)
df_dic
```

```python
# Bonus dengan lambda
df_dic['Bonus'] = df_dic.apply(lambda x: x['Gaji'] * 0.1, axis=1)
df_dic
```

```python
# Pajak dengan fungsi
def ppn(x):
    pajak= x['Gaji']
    kategori = x['Kategori']
    if kategori == "Muda":
        return pajak * 0.05
    else:
        return pajak * 0.1

df_dic['Pajak'] = df_dic.apply(ppn, axis=1)
df_dic
```

```python
# Pajak dengan lambda
df_dic['Pajak'] = df_dic.apply(lambda x: x['Gaji'] * 0.05 if x['Kategori'] == "Muda" else x['Gaji'] * 0.1, axis=1)
df_dic
```
